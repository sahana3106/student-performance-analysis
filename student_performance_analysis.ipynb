```python
y=df['overall_score'].head(30).values
```

```python
plt.scatter(x,y,color='red',alpha=0.7)
plt.grid(color='lightgrey')
plt.xlabel("hours spent on studying",fontsize=20,color='b')
plt.ylabel("overall marks",fontsize=20,color='b')
plt.title("Analysis on study hours",fontsize=25)

plt.show()
```

```python
series_so=df.groupby('study_method')['overall_score'].mean()
```

```python
x=series_so.index
y=series_so.values
plt.barh(x,y,color='pink')
plt.title("Analysis on study method",fontsize=20)
plt.xlabel("overall_marks",color='b',fontsize=15)
plt.ylabel("study method",color='b',fontsize=15)
plt.grid()
plt.show()
```

```python
#what: from the above bar graph we can say that the there is similar trend for all type of 
#studying with slight variation
#why:this is maybe because the study method are just a source to study and the 
#overall score depends on the student potential and understanding
#what so:this suggests that study method doesnt make any significant change in the overall scores
```

```python
series_po=df.groupby('parent_education')['overall_score'].mean()
```

```python
x=series_po.index
y=series_po.values
plt.bar(x,y,color='orange')
plt.tick_params(axis='both',color='r')
plt.title("analysis on parents education",fontsize=20)
plt.xlabel("parents education",color='r',fontsize=15)
plt.ylabel("overall_score",color='r',fontsize=15)
plt.grid()
plt.show()
```

```python
#Parent Education vs Score
#what:the average score is similar for all parent education type with slight difference in them.
#this maybe because the parents education make slight difference on student performance.
#students overall score is affect by other factors.
#what so:this suggests that parents education doesnt make any huge impact for study's overall score
```
